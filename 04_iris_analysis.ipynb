# Iris Dataset: Exploration and Visualization

```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_csv("data/IRIS.csv")
df
```

```python
print("Columns:", list(df.columns))
print("Shape:", df.shape)
```

```python
df.info()
```

```python
df.describe()
```

```python
df["sepal_length"].value_counts()
```

```python
features = ["sepal_length", "sepal_width", "petal_length", "petal_width"]

for col in features:
    df[col].plot(kind="hist", edgecolor="black", linewidth=1)
    plt.xlabel(col)
    plt.title(f"Histogram of {col}")
    plt.show()
```

```python
for col in features:
    df[col].plot(kind="box")
    plt.title(f"Box Plot of {col}")
    plt.show()
```

```python
df.plot(x="sepal_length", y="sepal_width", kind="scatter")
plt.title("Sepal Length vs Sepal Width")
plt.show()

df.plot(x="petal_length", y="petal_width", kind="scatter")
plt.title("Petal Length vs Petal Width")
plt.show()
```

```python
df["species"].value_counts().plot(kind="bar", edgecolor="black")
plt.title("Species Count")
plt.show()
```
