# Data Visualization: Students Performance

```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_csv("data/exams.csv")
df.head()
```

```python
print("Shape:", df.shape)
print("Duplicate rows:", df.duplicated().sum())
```

```python
counts = df["race/ethnicity"].value_counts()
plt.pie(counts, labels=counts.index, autopct="%1.1f%%")
plt.title("Race/Ethnicity Distribution")
plt.show()
```

```python
plt.scatter(df["math score"], df["reading score"], alpha=0.6)
plt.xlabel("Math score")
plt.ylabel("Reading score")
plt.title("Math vs Reading Score")
plt.show()
```

```python
df["math score"].plot(kind="box")
plt.title("Math Score Box Plot")
plt.show()
```

```python
df["math score"].plot(kind="hist", edgecolor="black")
plt.xlabel("Math score")
plt.title("Math Score Histogram")
plt.show()
```
