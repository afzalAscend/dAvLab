# NumPy and Pandas: Indexing and Slicing

```python
import numpy as np
import pandas as pd
```

```python
data = np.array(
    [
        [101, "Alice", 85],
        [102, "Bob", 90],
        [103, "Charlie", 78],
        [104, "David", 92],
        [105, "Emma", 88],
    ]
)

df = pd.DataFrame(data, columns=["ID", "Name", "Marks"])
print("Original DataFrame:")
print(df)
```

```python
# Indexing
print("\nName column:")
print(df["Name"])

print("\nName and Marks columns:")
print(df[["Name", "Marks"]])

print("\nFirst row:")
print(df.iloc[0])

print("\nMarks of Alice:")
print(df.iloc[0, 2])
```

```python
# Slicing
print("\nRows 1 to 3:")
print(df.iloc[1:4])

print("\nFirst three rows:")
print(df.iloc[:3])

print("\nLast two rows:")
print(df.iloc[-2:])

print("\nRows 1 to 3 and columns Name to Marks:")
print(df.iloc[1:4, 1:3])
```
